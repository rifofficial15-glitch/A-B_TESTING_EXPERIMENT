```python
%pip install statsmodels
from statsmodels.stats.proportion import proportions_ztest

# EXPERIMENT DATA
control_satisfied = 7500
control_total = 10000
treatment_satisfied = 7800
treatment_total = 10000

# NUMBER OF SATISFIED USERS
count = [treatment_satisfied, control_satisfied]

# TOTAL USERS IN EACH GROUP
nobs = [treatment_total, control_total]

# PERFORM TWO-PROPORTION Z-TEST
z_stat, p_value = proportions_ztest(count, nobs, alternative = 'larger')
'''HERE WE USE ALTERNATIVE = 'larger', BECAUSE OUR SPECIFIC HYPOTHESIS IS treatment_satisfaction > control_satisfaction'''
print("Z-STATISTIC:", z_stat)
print("P_VALUE:", p_value)

# DECISION
if p_value < 0.05:
    print("REJECT NULL HYPOTHESIS")
    print("NEW ALGORITHM IMPROVES USER SATISFACTION")
else:
    print("FAILED TO REJECT NULL HYPOTHESIS")
    print("NO STATISTICALLY SIGNIFICANT IMPROVEMENT")
    
```

```python
                        # CALCULATE THE CONFIDENCE INTERVAL

''' HOW MUCH IMPROVEMENT CAN WE REASONABLY EXPECT?'''

from statsmodels.stats.proportion import(confint_proportions_2indep)

# CALCULATE 95% CONFIDENCE INTERVAL

lower, upper = confint_proportions_2indep(treatment_satisfied, treatment_total, control_satisfied, control_total, method = 'newcomb')
print("LOWER LIMIT:", lower)
print("UPPER LIMIT:", upper)
```

```python
                        # CHECK STATISTICAL POWER
''' STATISTICAL POWER IS THE PROBABILITY OF DETECTING THE REAL EFFECT '''

from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

# EXPECTED SATISFACTION RATES
control_rate = 0.75
treatment_rate = 0.77

# CALCULATE EFFECT SIZE
effect_size = proportion_effectsize(treatment_rate, control_rate)
print("EFFECT SIZE:", effect_size)
# CALCULATE STATISTICAL POWER
power_analysis = NormalIndPower()
power = power_analysis.power(effect_size=effect_size,nobs1=10000,alpha=0.05,ratio=1, alternative='larger')
print("STATISTICAL POWER:",power)
```

```python
                # A. CTR (CLICK THROUGH RATE)

from statsmodels.stats.proportion import proportions_ztest

# CLICK & IMPRESSION COUNTS
control_clicks = 6000
treatment_clicks = 6300

control_impressions = 10000
treatment_impressions = 10000

count = [treatment_clicks, control_clicks]
nobs = [treatment_impressions, control_impressions]

z_stat, p_value = proportions_ztest(count, nobs, alternative="two-sided")

print("CTR CONTROL:", control_clicks / control_impressions)
print("CTR TREATMENT:", treatment_clicks / treatment_impressions)
print("Z STATISTICS:", z_stat)
print("P VALUE:", p_value)
```

CONCLUSION : 
        THE NEW ALGORITHM SIGNIFICANTLY INCREASED CLICK-THROUGH RATE..

```python
                                        # B. LONG - CLICK RATE

''' IT MEANS A USER CLICKS A SEARCH RESULT AND SPENDS A MEANINGFUL AMUNT OF TIME ON THE DESTINATION PAGE '''

control_Longclicks = 4500
treatment_Longclicks = 4900

control_total = 10000
treatment_total = 10000

count = [treatment_Longclicks, control_Longclicks]
nobs = [treatment_total, control_total]

z_stat, p_value = proportions_ztest(count, nobs, alternative="two-sided")

print("LONG-CLICK Z-STAT:", z_stat)
print("LONG-CLICK P-VALUE:", p_value)
```

CONCLUSION:

        LONG-CLICK RATE INCREASED FROM 45% TO 49% UNDER THE NEW ALGORITHM..

```python
                            # C. QUERY REFORMULATION RATE
''' QUERY REFORMULATION HAPPENS WHEN USERS CHANGE THEIR SEARCH QUERY BECAUSE THE ORIGINAL RESULTS WERE NOT USEFUL '''

control_reform = 2000
treatment_reform = 1600

count = [treatment_reform, control_reform] 
nobs = [10000, 10000]

z_stat, p_value = proportions_ztest(count, nobs, alternative="two-sided")

print("Z-STATISTICS:", z_stat)
print("P-VALUE:", p_value)
```

```python
                            # A.SEARCH LATENCY
''' SEARCH LATENCY MEASURES HOW LONG THE SEARCH ENGINE TAKES TO RETURN RESULTS '''

import numpy as np 

from scipy.stats import ttest_ind
np.random.seed(42)

# SIMULATED SEARCH LATENCY IN MILLISECONDS

control_latency = np.random.normal(loc=200, scale=40, size=10000)
treatment_latency = np.random.normal(loc=230, scale=45, size=10000)

# WELCH'S T-TEST

t_stat, p_value = ttest_ind(treatment_latency, control_latency, equal_var=False)

print("CONTROL MEAN:", np.mean(control_latency))
print("TREATMENT MEAN:", np.mean(treatment_latency))
print("T STATISTICS:", t_stat)
print("P VALUE:", p_value)
```

CONCLUSION:

        THE SEARCH LATENCY INCREASED SIGNIFICANTLY.. SO THIS GUARDRAIL METRIC FAILS( OR ATLEAST NEEDS ATTENTION)

```python
                            # B.ERROR RATE
''' ERROR RATE MEASURES HOW FREQUENTLY TECHINICAL ERROR OCCURS '''

control_errors = 50
treatment_errors = 50

control_total = 10000
treatment_total = 10000

count = [treatment_errors, control_errors]
nobs = [treatment_total, control_total]

z_stat, p_value = proportions_ztest(count, nobs, alternative="two-sided")

print("Z STATISTICS:", z_stat)
print("P VALUE:", p_value) 
```

CONCLUSION:

        ERROR RATE WAS UNCHANGED.. SO NO RELIABILLITY DEGRADATION WAS OBSERVED..

```python
                            # C.PAGE_LOADING_TIME
''' PAGE LOADING TIME MEASURES HOW LONG THE SEARCH PAGE TAKES TO BECOME USABLE'''
''' CONTROL:1.5 seconds; TREATMENT: 1.7 seconds '''

control_load_time = np.random.normal(loc=1.5, scale=0.3, size=10000)
treatment_load_time = np.random.normal(loc=1.7, scale=0.35, size=10000)

t_stat, p_value = ttest_ind(treatment_load_time, control_load_time, equal_var=False)

print("CONTROL MEAN:", np.mean(control_load_time))
print("TREATMENT MEAN:", np.mean(treatment_load_time))
print("T STATISTICS:", t_stat)
print("P VALUE:", p_value)
```

CONCLUSION:

        MEAN PAGE LOADING TIME INCREASED SIGNIFICANTLY UNDER THE NEW ALGORITHM, SO THIS GUARDRAIL METRIC IS BREACHED
